# Can Chi calendar 

```python
def calculate_can_chi_calendar(year):
    result = ''

    can = int(year)%10
    chi = int(year)%12
    
    match can:
        case 0:
            result += 'Canh '
        case 1:
            result += 'Tân '
        case 2:
            result += 'Nhâm '
        case 3:
            result += 'Quý '
        case 4:
            result += 'Giáp '
        case 5:
            result += 'Ất '
        case 6:
            result += 'Bính '
        case 7:
            result += 'Đinh '
        case 8:
            result += 'Mậu '
        case 9:
            result += 'Kỷ '

    match chi:
        case 0:
            result += 'Thân'
        case 1:
            result += 'Dậu'
        case 2:
            result += 'Tuất'
        case 3:
            result += 'Hợi'
        case 4:
            result += 'Tý'
        case 5:
            result += 'Sửu'
        case 6:
            result += 'Dần'
        case 7:
            result += 'Mẹo'
        case 8:
            result += 'Thìn'
        case 9:
            result += 'Tỵ'
        case 10:
            result += 'Ngọ'
        case 11:
            result += 'Mùi'
    
    return result 
```

```python
calculate_can_chi_calendar(1988)
```
